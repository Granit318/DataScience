```python
import numpy as np
```

```python

def approach(expr, points):
    for x in points:
        print(float(x), float(expr(x)))


def check(f, df, x, tol=1e-4):
    got = df(x)
    if got is None:
        print("производная ещё не написана")
        return
    h = 1e-5
    num = (f(x + h) - f(x - h)) / (2 * h)
    gap = abs(float(got) - float(num))
    print("ваша", float(got), "численная", float(num), "ok" if gap < tol else "проверьте формулу")
```

### 5. Неявная производная

Три уравнения со слайда. Допишите `yp(x, y)`. Ячейка проверяет формулу в одной точке кривой: сравнивает её с численным наклоном. Точку не меняйте.


```python
def numerical_slope(F, x, y, h=1e-5):
    def solve(x1, y0):
        yy = float(y0)
        for _ in range(30):
            d = (F(x1, yy + 1e-6) - F(x1, yy - 1e-6)) / 2e-6
            val = F(x1, yy)
            if abs(d) < 1e-12:
                break
            yy = yy - val / d
        return yy
    return (solve(x + h, y) - solve(x - h, y)) / (2 * h)

def check_implicit(F, yp, x, y):
    got = yp(x, y)
    if got is None:
        print('формула ещё не написана')
        return
    num = numerical_slope(F, x, y)
    gap = abs(float(got) - float(num))
    print('ваша', float(got), 'численная', float(num), 'ok' if gap < 1e-4 else 'проверьте формулу')
    print('уравнение в точке', float(F(x, y)))
```

#### 5.1. 2x³ − 5xy − y² = 3


```python
def F(x, y):
    return 2 * x**3 - 5 * x * y - y**2 - 3

def yp(x, y):
    return None

x, y = 1.0, (-5 + np.sqrt(21)) / 2
check_implicit(F, yp, x, y)
```

#### 5.2. x² − 3xy + y² = 1


```python
def F(x, y):
    return x**2 - 3 * x * y + y**2 - 1

def yp(x, y):
    return None

x, y = 0.0, 1.0
check_implicit(F, yp, x, y)
```

#### 5.3. y³ − 3y² − 9y + 2x² + 20x = −27
